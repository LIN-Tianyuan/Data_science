```python
jump_distance = 5
distances = [1, 3, 4, 2, 1, 4, 5]
n = len(distances)
jumps = 0
jump_remaining = 0
go = True
i = 0

while go:
    if i>=n:
        go = False
        print('End')
        break
    else:
        print(i)

    if distances[i] <= jump_remaining:
        jumps += 1
        jump_remaining -= distances[i]
        print('continue jump another', distances[i])
        i += 1
    elif distances[i] <= jump_distance:
        jumps += 1
        jump_remaining = jump_distance
        jump_remaining -= distances[i]
        print('jump', distances[i])
        i += 1
    else:
        print('Too far')
        go = False
```

```python
from itertools import accumulate

def frog(distances, n):
    # Part a: every gap must be at most n
    if any(d > n for d in distances):
        return False, None

    positions = [0] + list(accumulate(distances))
    last = len(positions) - 1

    # Part b: always jump to the furthest point within n centimetres
    current = 0
    jumps = 0
    while current != last:
        nxt = current
        while nxt < last and positions[nxt + 1] - positions[current] <= n:
            nxt += 1
        current = nxt
        jumps += 1

    return True, jumps

reachable, jumps = frog([3, 5, 2, 4, 6, 4], 6)
print(reachable, jumps) # True 5
```
