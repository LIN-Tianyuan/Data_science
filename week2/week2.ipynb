```python
# 4. Extract one of the columns and store it as a list
countries = data['Name']
countries
```

```python
populations = data['Population']
populations
```

```python
# 6. Work out the average of populations
total = 0
for pop in populations:
    total += pop

avg = total / len(populations)
print('Average:', avg)
```

```python
# 7. Put this code inside a function, and call it to work out the average again
def calculate_average(my_list):
    total = 0
    for i in my_list:
        total += i
    avg = total / len(my_list)
    return avg

avg = calculate_average(populations)
print('Average:', avg)
```
